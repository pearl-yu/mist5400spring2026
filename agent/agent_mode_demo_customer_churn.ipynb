## What is This Notebook?

Each section below is a prompt you will **copy and paste into the Gemini agent panel** in Google Colab (or type it in your own words). The agent will write and run the code for you.

**Your job is to:**
1. Understand what each prompt is asking the agent to do and why
2. Review the agent's output before moving on
3. Give feedback or ask follow-up questions when something doesn't look right

This is how agentic AI works: you set the goal, the agent plans and executes, and you evaluate and steer.

## The Task

**Business scenario:** You are a data analyst at ShopStream, a mid-size e-commerce company. Leadership is concerned about customer churn — customers who stop buying. They want you to:

1. Explore the customer data and surface key patterns
2. Build a predictive model: which customers are likely to churn?
3. Identify the most important drivers of churn
4. Provide actionable recommendations for the retention team

**Dataset:** `ecommerce_customer_churn.csv` — 1,000 customers with 13 columns including demographics, behavior, satisfaction, and whether they churned.

## The Dataset: Column Descriptions

| Column | Description |
|--------|-------------|
| `customer_id` | Unique identifier for each customer |
| `age` | Customer's age |
| `gender` | Male, Female, or Non-Binary |
| `tenure_months` | How long they've been a customer (in months) |
| `monthly_spend` | Average monthly spending in dollars |
| `num_support_tickets` | Number of support tickets filed |
| `payment_delay_days` | Average number of days payments are delayed |
| `last_login_days_ago` | Days since the customer last logged in |
| `num_products_purchased` | Total products purchased |
| `satisfaction_score` | Self-reported satisfaction (1.0 to 5.0) |
| `contract_type` | Month-to-Month, Annual, or Two-Year |
| `uses_mobile_app` | Whether the customer uses the mobile app (0 or 1) |
| `churned` | **Target variable** — 1 if the customer churned, 0 if retained |

## Prompt 1: Set the Background

> **Why this matters:** Before giving the agent tasks, you set the context — just like briefing a new team member. This is the "background" component of the agent mode workflow from class.

Copy this into the Gemini agent panel:

---

```
You are an ML project copilot helping me with a business analytics project.

I'm a data analyst at ShopStream, an e-commerce company. We're losing customers
and leadership wants to understand why and predict who will churn next.

I've uploaded a dataset called ecommerce_customer_churn.csv with 1,000 customers.
The target variable is "churned" (1 = churned, 0 = retained).

Throughout this project:
- Always explain what you're doing and why before writing code
- Use clear visualizations with titles and labels
- After each step, suggest what we should do next
- Use plain business language first, then technical details
```

---

**What to look for in the agent's response:** It should acknowledge the scenario and likely offer to start by loading and exploring the data. If it jumps straight into modeling, redirect it — always start with exploration.

BTW, we could further **customize the instructions via the configure icon** in the Gemini panel.

## Prompt 2: Load and Explore the Data

> **Why this matters:** The agent needs to "see" the data before it can reason about it. This fills its context window with real information about your dataset — column types, missing values, distributions.

You can add the dataset as a context into the Gemini panel.

---

```
Please load the dataset ecommerce_customer_churn.csv.

Show me:
- The first few rows so I can verify it loaded correctly
- The shape of the dataset (rows and columns)
- Data types for each column
- Check for any missing values
- Basic descriptive statistics for the numeric columns
```

---

**What to look for:** Verify the agent loaded all 1,000 rows and 13 columns. Check that it correctly identifies `churned` as the target. If anything looks off, tell it.

## Prompt 3: Understand the Target Variable

> **Why this matters:** Before modeling, you need to understand what you're predicting. Is the dataset balanced? A heavily imbalanced target will affect model choice and evaluation.

---

```
Now let's understand our target variable.

Show me the churn rate — what percentage of customers churned vs. were retained?
Visualize this with a bar chart.

Also, are there any interesting patterns? For example:
- Do customers on Month-to-Month contracts churn more?
- Is satisfaction score different for churned vs. retained customers?
- How about last login days — do churned customers log in less often?

Show me 3-4 visualizations that compare churned vs. retained customers
across the most interesting features.
```

---

**What to look for:** The visualizations should clearly show differences between churned and retained groups. If a chart is hard to read, ask the agent to fix it: *"That chart is hard to read. Can you use side-by-side box plots instead?"*

**This is an important skill** — giving the agent feedback and iterating is exactly how agentic workflows function.

## Prompt 4: Dig Deeper — Correlations

> **Why this matters:** Understanding which features relate to churn helps you make smarter modeling decisions and gives you early hypotheses to share with the business team.

---

```
Create a correlation heatmap for all the numeric features, including the
target variable "churned".

Which features have the strongest correlation with churn?
Summarize the top 3-4 findings in plain English.
```

---

**What to look for:** The agent should produce a clean heatmap and translate the numbers into business insights, such as *"Customers with lower satisfaction scores are more likely to churn."*

## Prompt 5: Prepare the Data for Modeling

> **Why this matters:** Models need numeric inputs and properly split data. This is where the agent handles encoding, splitting, and any necessary preprocessing.

---

```
Now let's prepare the data for modeling.

- Drop the customer_id column (it's just an identifier, not a feature)
- Encode any categorical variables (gender, contract_type) appropriately
- Split the data into training (80%) and testing (20%) sets
- Use a random seed of 42 so the results are reproducible

Tell me the size of the training and test sets when you're done.
```

---

**What to look for:** Confirm 800 training rows and 200 test rows. Make sure `customer_id` was dropped and categorical variables were encoded. If the agent uses one-hot encoding vs. label encoding, that's a good thing to notice and ask about.

## Prompt 6: Build the First Model — Logistic Regression

> **Why this matters:** Always start simple. Logistic regression gives you an interpretable baseline. From class: *"Always start with a simple model."*

---

```
Let's start with a simple model.

Build a Logistic Regression to predict churn using the training data.
Evaluate it on the test set and show me:
- Accuracy
- Precision, Recall, and F1-score (especially for the "churned" class)
- A confusion matrix

Explain the results in business terms: if we used this model to identify
at-risk customers, how well would it do?
```

---

**What to look for:** Pay attention to recall for the churned class — that's the model's ability to catch customers who are about to leave. High accuracy but low recall means the model misses many churners, which is bad for the business goal.

## Prompt 7: Build a Second Model — Random Forest

> **Why this matters:** Comparing models is fundamental. A more complex model should earn its complexity by performing noticeably better.

---

```
Now build a Random Forest classifier on the same training data.

Show me the same evaluation metrics: accuracy, precision, recall, F1-score,
and confusion matrix.

Compare the Random Forest results with the Logistic Regression.
Which model is better at catching churners? Which one makes fewer
false alarms? Summarize the trade-offs.
```

---

**What to look for:** The agent should give you a clear comparison. If both models are close, that's fine. If the Random Forest is significantly better, ask *why* — what is it capturing that logistic regression missed?

## Prompt 8: Ask the Agent for Improvement Ideas

> **Why this matters:** This is where the agentic workflow shines — you don't need to know every technique. You ask the agent to reason about what might help, then you decide whether to try it.

---

```
Based on the results so far, what do you suggest to improve the model's
performance?

Give me 2-3 concrete suggestions. For each one, explain:
- What it does
- Why it might help
- How much effort it would take

Don't implement anything yet — just suggest. I'll tell you which one
to try.
```

---

**This is a key agentic interaction pattern:** You ask the agent to *suggest* rather than *act*, review the options, then give the go-ahead. This keeps you in the driver's seat.

After reviewing the suggestions, pick one and tell the agent:

```
Let's try [the suggestion you picked]. Go ahead and implement it
and show me the updated results compared to our previous models.
```

## Prompt 9: Feature Importance — What Drives Churn?

> **Why this matters:** The business team doesn't care about model accuracy alone — they want to know *what to do about it*. Feature importance translates your model into action.

---

```
Using the best model we've built so far, show me the feature importance.

Create a horizontal bar chart ranking features from most to least important.

For the top 5 most important features, explain in business terms:
- What does this feature tell us about why customers leave?
- What could the retention team do about it?
```

---

**What to look for:** The agent should connect data insights to business actions. For example, if `satisfaction_score` is the top driver, the recommendation might be to invest in customer experience improvements.

## Prompt 10: Generate the Executive Summary

> **Why this matters:** Every analytics project should end with a clear summary for stakeholders who won't read your code. The agent can draft this for you.

---

```
Please write a concise executive summary of this entire analysis.
Structure it as:

1. Business Problem (2-3 sentences)
2. What We Found in the Data (3-4 key findings from exploration)
3. Model Performance (which model we chose and how well it performs)
4. Top Drivers of Churn (the most important features)
5. Recommendations (3-4 specific actions for the retention team)

Keep it under 300 words. Use plain business language, no jargon.
```

---

**What to look for:** Read the summary critically. Does it accurately reflect what the data showed? Would a non-technical executive understand it? If not, give feedback: *"The second point is too technical — rewrite it for a non-technical audience."*

## Prompt 11: Final Visualization — Model Comparison

> **Why this matters:** A side-by-side visual comparison is the best way to justify your model choice to stakeholders.

---

```
Create a single visualization that compares all the models we built.

Show accuracy, precision, recall, and F1-score for each model
side by side using a grouped bar chart.

Title it: "Model Performance Comparison — Customer Churn Prediction"
Make it presentation-ready.
```

---

**What to look for:** The chart should be clean and readable at a glance. If it's cluttered, ask the agent to simplify it.
